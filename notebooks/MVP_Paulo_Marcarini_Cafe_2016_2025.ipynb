# 1. Problema e perguntas

O café é um dos principais produtos da pauta exportadora brasileira e possui relevância econômica tanto pela geração de divisas quanto pela inserção do Brasil no mercado internacional de commodities agrícolas.

A compreensão do comportamento das exportações ao longo do tempo pode auxiliar na identificação de padrões de volume, valor comercial e distribuição geográfica dos embarques.

Este trabalho utiliza dados do [Comex Stat](https://comexstat.mdic.gov.br/pt/home), disponibilizados pelo Ministério do Desenvolvimento, Indústria, Comércio e Serviços (MDIC), no seu [portal de Estatísticas de Comércio Exterior em Dados Abertos](https://www.gov.br/mdic/pt-br/assuntos/comercio-exterior/estatisticas/base-de-dados-bruta), para analisar as exportações brasileiras de café não torrado, não descafeinado, em grão (NCM 09011110) realizadas entre janeiro de 2016 e dezembro de 2025.

Cada registro da base representa uma observação agregada de exportação para uma combinação específica de período, país de destino, unidade federativa exportadora, unidade aduaneira e demais características logísticas associadas ao embarque.

## 1.1 Contexto e objetivo


O objetivo analítico deste MVP é realizar uma análise exploratória e um pré-processamento dos dados de exportação de café em grão, buscando identificar padrões temporais e comportamentos associados ao volume e ao valor comercial das exportações brasileiras ao longo do período analisado.


## 1.2 Perguntas ou hipóteses

- **H1:** As exportações brasileiras de café em grão apresentam comportamento sazonal ao longo dos meses do ano.

- **H2:** O valor FOB por quilograma**¹** das exportações brasileiras de café em grão variou ao longo do período de 2016 a 2025.



**Premissas e restrições:**

- O estudo está restrito à NCM 09011110 (**café não torrado, não descafeinado, em grão**).

- A análise utiliza exclusivamente informações do **Comex Stat**.

- O trabalho possui caráter descritivo e exploratório, não buscando estabelecer relações de causalidade.

- Não serão utilizados modelos de aprendizado de máquina.

- Eventuais mudanças observadas ao longo do período podem estar associadas a diversos fatores externos não presentes na base, como preços internacionais, câmbio, condições climáticas e dinâmica da demanda global.


---
<small>**Nota explicativa:**</small>


<small>**¹ FOB por quilograma (US$/kg):** é o preço líquido, em dólares americanos (valores nominais), de 1 kg de café brasileiro já colocado dentro do navio no Brasil, pronto para viajar para outro país, sem incluir os custos do transporte internacional e do seguro da viagem.</small>
